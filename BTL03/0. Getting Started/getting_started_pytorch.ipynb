# Import PyTorch

```python
import torch
import torchvision
```

# Check PyTorch & CUDA version

```python
print('PyTorch version:', torch.__version__)
print('CUDA available:', torch.cuda.is_available())
print('CUDA version (PyTorch build):', torch.version.cuda)
```

# Check whether GPU can be used or not

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Selected device:', device)

if torch.cuda.is_available():
    num_gpus = torch.cuda.device_count()
    print(f'Number of GPUs detected: {num_gpus}')
    for i in range(num_gpus):
        props = torch.cuda.get_device_properties(i)
        vram_gb = props.total_memory / (1024 ** 3)
        print(f'  GPU {i}: {props.name} | VRAM: {vram_gb:.2f} GB | Compute Capability: {props.major}.{props.minor}')
else:
    print('No GPU detected. Running on CPU.')
```

## Quick Tensor Test on Selected Device

```python
x = torch.randn(3, 3, device=device)
y = torch.matmul(x, x.T)
print('Tensor device:', y.device)
print(y)
```
